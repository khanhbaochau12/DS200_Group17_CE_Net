```python
import os
import cv2
import numpy as np
```

```python
def convert_images(input_dir, output_dir):
    os.makedirs(output_dir, exist_ok=True)

    for file in os.listdir(input_dir):
        if file.lower().endswith((".tif", ".tiff")):
            in_path = os.path.join(input_dir, file)
            out_path = os.path.join(output_dir, file.replace(".tif", ".png"))

            # đọc ảnh grayscale
            img = cv2.imread(in_path, cv2.IMREAD_GRAYSCALE)

            if img is not None:
                cv2.imwrite(out_path, img)
                print(f"Image done: {file}")
            else:
                print(f"Error: {file}")
```

```python
def convert_masks(input_dir, output_dir):
    os.makedirs(output_dir, exist_ok=True)

    for file in os.listdir(input_dir):
        if file.lower().endswith((".tif", ".tiff")):
            in_path = os.path.join(input_dir, file)
            out_path = os.path.join(output_dir, file.replace(".tif", ".png"))

            # đọc mask
            mask = cv2.imread(in_path, cv2.IMREAD_GRAYSCALE)

            if mask is not None:
                # đảm bảo binary đúng nghĩa (0 và 255)
                mask = (mask > 0).astype(np.uint8) * 255

                cv2.imwrite(out_path, mask)
                print(f"Mask done: {file}")
            else:
                print(f"Error: {file}")
```

```python
base = "Datasets/LUNA"

convert_images(
    input_dir=os.path.join(base, "2d_images"),
    output_dir=os.path.join(base, "2d_images_png")
)

convert_masks(
    input_dir=os.path.join(base, "2d_masks"),
    output_dir=os.path.join(base, "2d_masks_png")
)
```

```python
def merge_masks(left_dir, right_dir, output_dir):
    os.makedirs(output_dir, exist_ok=True)

    left_files = sorted(os.listdir(left_dir))

    for file in left_files:
        if file.lower().endswith(".png"):
            left_path = os.path.join(left_dir, file)
            right_path = os.path.join(right_dir, file)

            if not os.path.exists(right_path):
                print(f"Missing right mask for {file}")
                continue

            left_mask = cv2.imread(left_path, cv2.IMREAD_GRAYSCALE)
            right_mask = cv2.imread(right_path, cv2.IMREAD_GRAYSCALE)

            if left_mask is None or right_mask is None:
                print(f"Error reading {file}")
                continue

            # đảm bảo binary
            left_mask = (left_mask > 0).astype(np.uint8)
            right_mask = (right_mask > 0).astype(np.uint8)

            # merge (OR)
            merged = np.logical_or(left_mask, right_mask).astype(np.uint8) * 255

            out_path = os.path.join(output_dir, file)
            cv2.imwrite(out_path, merged)

            print(f"Merged: {file}")
```

```python
base = "Datasets/Montgomery/MontgomerySet"

merge_masks(
    left_dir=os.path.join(base, "ManualMask", "leftMask"),
    right_dir=os.path.join(base, "ManualMask", "rightMask"),
    output_dir=os.path.join(base, "ManualMask", "mergedMask")
)
```
